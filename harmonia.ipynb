```python
import pandas as pd
import ta
```

```python
harmonia_2022 = pd.read_csv("https://github.com/blockhain-elite-labs/harmonia-v20.1-82-backtests/raw/main/harmonia-20.1-82-2022-market-23months.csv.zip")
harmonia_2022["open_date"] = pd.to_datetime(harmonia_2022["open_date"])
```

```python
# indicators
window_slow = 26
window_fast = 9
harmonia_2022["MACD"] = ta.trend.macd(harmonia_2022["close"], window_slow=window_slow, window_fast=window_fast, fillna=False)
harmonia_2022["MACD_DIFF"] = ta.trend.macd_diff(harmonia_2022["close"], window_slow=window_slow, window_fast=window_fast, fillna=False)
harmonia_2022["EMA_MACD"] = ta.trend.macd_signal(harmonia_2022["close"], window_slow=window_slow, window_fast=window_fast, fillna=False)
harmonia_2022["MACD_DIFF"] = harmonia_2022["MACD"] - harmonia_2022["EMA_MACD"]
```
