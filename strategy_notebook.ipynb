```python
from ta.trend import SMAIndicator, EMAIndicator
def zlema(data, period):
    ema_data = data + (data - data.shift((period - 1) // 2))
    return EMAIndicator(ema_data, window=period).ema_indicator()
```

```python
x = 15
zlema_x = zlema(df["open"], period=x)
ma_x = SMAIndicator(df["open"], window=x).sma_indicator()
ema_x = EMAIndicator(df["open"], window=x).ema_indicator()
multiple_signals_same_graph(df["open_date"].values,{f"zlema_{x}":zlema_x.values,
                                                    f"ma_{x}":ma_x.values,
                                                    f"ema_{x}":ema_x.values,
                                                    "open":df["open"].values})
```
